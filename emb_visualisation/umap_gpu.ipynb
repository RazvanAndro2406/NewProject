```python
import itertools
from umap_functions import generate_umap_plot

embedding_paths = [
    #r"/bigdata/md5_s6cfa268729ee2148ee90bbb3acd/NewProject/embeddings_sibiu/embeddings_lstm_sibiu_0.txt",
    r"/bigdata/md5_s6cfa268729ee2148ee90bbb3acd/NewProject/embeddings_sibiu/embeddings_lstm_sibiu_1.txt",
    #r"/bigdata/md5_s6cfa268729ee2148ee90bbb3acd/NewProject/embeddings_sibiu/embeddings_lstm_sibiu_2.txt",
    r"/bigdata/md5_s6cfa268729ee2148ee90bbb3acd/NewProject/embeddings_sibiu/embeddings_lstm_sibiu_3.txt"
]

labels_paths = [
    #r"/bigdata/md5_s6cfa268729ee2148ee90bbb3acd/NewProject/labels_msl/msl_classes_990.csv",
    r"/bigdata/md5_s6cfa268729ee2148ee90bbb3acd/NewProject/labels_msl/msl_classes_995.csv",
    r"/bigdata/md5_s6cfa268729ee2148ee90bbb3acd/NewProject/labels_msl/msl_classes_1000.csv"
]

n_neighbors_list = [15]
n_components_list = [2] 
min_dist_list = [0.01, 0.1, 0.5]

grid = list(itertools.product(
    embedding_paths, 
    labels_paths, 
    n_neighbors_list, 
    n_components_list, 
    min_dist_list
))

print(f"Starting grid search: {len(grid)} total combinations to run.\n")

for emb_path, lbl_path, n_neigh, n_comp, m_dist in grid:
    emb_name = emb_path.split('/')[-1].split('\\')[-1]
    lbl_name = lbl_path.split('/')[-1].split('\\')[-1]
    
    print(f"{'-'*40}")
    print(f"Testing: {emb_name} + {lbl_name}")
    print(f"Params: neighbors={n_neigh}, components={n_comp}, min_dist={m_dist}")
    
    try:
        generate_umap_plot(
            embedding_path=emb_path, 
            labels_path=lbl_path, 
            n_neighbors=n_neigh,
            n_components=n_comp, 
            min_dist=m_dist
        )
    except Exception as e:
        print(f"ERROR on this combination: {e}")
```
